```python
f=open('i5_4300U.txt','r')
i5_4300U=f.read()
f.close()
f=open('i5_7300HQ.txt','r')
i5_7300HQ=f.read()
f.close()
f=open('GTX_1060.txt','r')
GTX_1060=f.read()
f.close()
```

```python
def listToFloat(l):
    list = l[:-1].split(',')
    for k in range(len(list)):
        list[k]=float(list[k])
    return list
def moyenne(l):
    return sum(l)/len(l)

def mediane(l):
    n=len(l)
    if n==2:
        return (l[n/2]+l[n/2+1])/2
    else:
        return l[int(n/2)+1]
```

```python
i5_4300U=listToFloat(i5_4300U)
i5_7300HQ=listToFloat(i5_7300HQ)
GTX_1060=listToFloat(GTX_1060)
```

```python
print(f'i5_4300U\nMoyenne: {moyenne(i5_4300U)}\nMediane: {mediane(i5_4300U)}')
print(f'i5 7300HQ\nMoyenne: {moyenne(i5_7300HQ)}\nMediane: {mediane(i5_7300HQ)}')
print(f'GTX 1060\nMoyenne: {moyenne(GTX_1060)}\nMediane: {mediane(GTX_1060)}')
```
